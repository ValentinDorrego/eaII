-  Realizando un análisis circuital, en el cual el nodo $V_{in}=0$ de la *Ilustración 2*, será denominado nodo $V_x$, determinamos una corriente que circula desde $V_{out}$ hacia $V_x$ a través de $R_f$, denominada $I$. Esta corriente se divide en dos componentes: una corriente $I_{R_s}$, que circula a través de $R_s$, y una pequeña corriente $I_{B1}$, correspondiente a la corriente de polarización que ingresa al terminal inversor del amplificador operacional. Asimismo, se considera la corriente de polarización $I_{B2}$, que ingresa al terminal no inversor, cuyo potencial es también $V_x$.

Por lo tanto, aplicando la Ley de Corrientes de Kirchhoff (LCK) en el nodo $V_x$, se obtiene:

$$
V_x = -I_{B2}\cdot R \\
$$
$$
I_{Rs} = \frac{V_x}{Rs}$$
$$
I = I_{R_s} + I_{B1} = \frac{V_{out}-V_x}{Rf}
$$
Operando un poco algebraicamente llegamos a
$$
\frac{V_{out}}{Rf} = V_x\left( \frac{1}{Rs} + \frac{1}{Rf}\right) + I_{B1}
$$
Si consideramos el caso en donde  $\left(\frac{1}{Rs} + \frac{1}{Rf}^{-1}\right) \approx R$, nos permite tener una diferencia de corrientes mucho menor en ambas entradas
$$
\frac{V_{out}}{Rf} = -I_{B2} + I_{B1} = I_{io}
$$
Y terminamos obteniendo lo que se denomina *tension de salida residual*

$$
\boxed{
V_{out}= I_{io} \cdot Rf}
$$

Teniendo esto en consideracion nos facilita el porqué de nuestros resultados.

 -   Como podemos observar, al mantener constante la relación $Rf/Rs$ en todos los casos, podemos analizar la influencia de la resistencia R sobre la corriente de salida. Para valores pequeños de R, la magnitud de la corriente de salida es relativamente elevada. A medida que R aumenta, la magnitud de dicha corriente disminuye hasta aproximarse a cero, y para valores mayores de R la corriente cambia de signo y vuelve a aumentar en magnitud.
    Este comportamiento se relaciona con la corriente de polarización de la entrada no inversora, ya que al aumentar R aumenta la caida de tension $Vz=−IB2R$, modificando las corrientes de polarizacion al offset de salida. Sabemos que $R=R//Rf$ permite igualar aproximadamente las resistencias vistas por ambas entradas y minimizar el efecto asociado a las corrientes de polarización. Sin embargo, el mínimo de la corriente de salida total observado en la simulacion nos coincide con este valor, puede que sea debido a que intervienen otros parametros del modelo del amplificador operacional, como la tension de offset de entrada.

- Como mencionamos anteriormente la reitencia $R$ permite poder igualar las resitencias vistas por ambas entradas del operacional, y minimizar la tension de sali residual, con $R = 0$ nos termina quedando la siguiente expresion:

$$
Vort = Vio (1 + \frac{R2}{R1}) + I_B\cdot Rs
$$

En donde $I_B \cdot Rs$ puede llegar a ser considerablemente mayor a los casos que analizamos, por lo tanto no poner la resistencia $R$ viene acompañado de un considerado mayor error de offset.

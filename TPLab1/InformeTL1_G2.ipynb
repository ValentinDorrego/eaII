## Corrección de errores estáticos: Corriente de polarización y offset de tensión

En esta primera sección se busca simular los errores estáticos del amplificador operacional y la
configuración como derivador estudiándolo desde un análisis de realimentación.
Para analizar los problemas estáticos del amplificador operacional se propone utilizar el CI LM741, una
alimentación simétrica de ±12V y una resistencia de carga de 10K.
Se deben simular las siguientes configuraciones, analizando cómo varía la tensión de offset modificando
el valor de R en el terminal no inversor, para distintos valores de Rs y Rf.

- Rs=10 - Rf=1K
- Rs=100 - Rf=10K
- Rs=1K - Rf=100K
- Rs=10K - Rf=1000K
